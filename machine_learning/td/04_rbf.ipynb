# TD04 — k plus proches voisins et réseau RBF

## Objectif

Deux modèles résolvent des données **non linéairement séparables** sans rétropropagation :

- **k-NN** (k plus proches voisins) : modèle *paresseux* — on ne s'entraîne pas, on répond en regardant les `k` points d'entraînement les plus proches de la requête (vote **dur**).
- **Réseau RBF** (fonctions à base radiale) : noyaux gaussiens centrés sur les points d'entraînement + couche de sortie linéaire. C'est un « k-NN adouci » : *tous* les voisins comptent, avec un poids qui décroît avec la distance (pondération **douce**).

Au programme :
1. Implémenter **k-NN** en numpy (distance + vote majoritaire).
2. Construire un **réseau RBF** en réutilisant la classe `MLP` du TD03 pour la couche de sortie.
3. **Comparer** leurs frontières de décision et le rôle de `k` et de `σ`.

> Le réseau RBF est un pont entre k-NN (partie 6.a du cours) et le perceptron (TD03) : sa sortie est un neurone linéaire entraîné sur les activations radiales $φ(x)$.

```python
import numpy as np
import matplotlib.pyplot as plt
import copy

# Alias de types pour les tableaux numpy
Vec = np.ndarray[tuple[int], np.dtype[np.float64]]        # vecteur 1D de flottants
Mat = np.ndarray[tuple[int, int], np.dtype[np.float64]]   # matrice 2D de flottants
```

## 1. Classe `MLP` réutilisée (TD03)

La couche de sortie du RBF est un simple neurone **linéaire + sigmoïde** sur les activations $φ(x)$. C'est exactement ce que fait la classe `MLP` du TD03 avec une seule couche. On la **réutilise telle quelle** (copiée de `mlp.ipynb`) : il suffira de lui donner `Phi(X)` en entrée.

> Seule `fit_batch` (descente de gradient par lot) nous sert ici.

```python
def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))


def sigmoid_prime(z):
    s = sigmoid(z)
    return s * (1.0 - s)


def tanh_prime(z):
    return 1.0 - np.tanh(z) ** 2


ACTIVATIONS = {
    "sigmoid": (sigmoid, sigmoid_prime),
    "tanh": (np.tanh, tanh_prime),
    "linear": (lambda z: z, lambda z: np.ones_like(z)),
}
```

```python
class MLP:
    def __init__(self, sizes, activation="sigmoid", output_activation="linear", seed=0):
        rng = np.random.default_rng(seed)
        self.sizes = list(sizes)
        self.Ws = []
        self.bs = []
        for n_in, n_out in zip(sizes[:-1], sizes[1:]):
            self.Ws.append(rng.normal(0.0, 0.5, size=(n_out, n_in)))
            self.bs.append(np.zeros(n_out))
        self.activation = activation
        self.output_activation = output_activation

    def predict(self, X):
        return self.forward(X)[0][-1]

    def loss(self, X, t):
        return 0.5 * float(np.mean(np.sum((self.predict(X) - t) ** 2, axis=1)))

    def _step(self, X, t, lr):
        for l, (dW, db) in enumerate(self.backward(X, t)):
            self.Ws[l] -= lr * dW
            self.bs[l] -= lr * db

    def forward(self, X):
        A = X
        As = [A]
        Zs = []
        n = len(self.Ws)
        for l, (W, b) in enumerate(zip(self.Ws, self.bs)):
            Z = A @ W.T + b
            Zs.append(Z)
            name = self.output_activation if l == n - 1 else self.activation
            A = ACTIVATIONS[name][0](Z)
            As.append(A)
        return As, Zs

    def backward(self, X, t):
        As, Zs = self.forward(X)
        N = X.shape[0]
        n = len(self.Ws)
        delta = (As[-1] - t) * ACTIVATIONS[self.output_activation][1](Zs[-1])
        dWs, dbs = [], []
        for l in range(n - 1, -1, -1):
            dW = delta.T @ As[l] / N
            db = delta.sum(axis=0) / N
            dWs.insert(0, dW)
            dbs.insert(0, db)
            if l > 0:
                delta = (delta @ self.Ws[l]) * ACTIVATIONS[self.activation][1](Zs[l - 1])
        return list(zip(dWs, dbs))

    def fit_batch(self, X, t, X_test=None, t_test=None, lr=0.1, epochs=2000, patience=200):
        history = {"train": [], "test": []}
        best_test = np.inf
        best = None
        no_improve = 0
        for epoch in range(epochs):
            self._step(X, t, lr)
            history["train"].append(self.loss(X, t))
            if X_test is not None:
                te = self.loss(X_test, t_test)
                history["test"].append(te)
                if te < best_test:
                    best_test = te
                    best = (copy.deepcopy(self.Ws), copy.deepcopy(self.bs))
                    no_improve = 0
                else:
                    no_improve += 1
                    if no_improve >= patience:
                        break
        if best is not None:
            self.Ws, self.bs = best
        return history

    def fit_online(self, X, t, X_test=None, t_test=None, lr=0.1, epochs=200, patience=50, seed=0):
        rng = np.random.default_rng(seed)
        history = {"train": [], "test": []}
        best_test = np.inf
        best = None
        no_improve = 0
        N = X.shape[0]
        for epoch in range(epochs):
            for i in rng.permutation(N):
                self._step(X[i:i + 1], t[i:i + 1], lr)
            history["train"].append(self.loss(X, t))
            if X_test is not None:
                te = self.loss(X_test, t_test)
                history["test"].append(te)
                if te < best_test:
                    best_test = te
                    best = (copy.deepcopy(self.Ws), copy.deepcopy(self.bs))
                    no_improve = 0
                else:
                    no_improve += 1
                    if no_improve >= patience:
                        break
        if best is not None:
            self.Ws, self.bs = best
        return history
```

```python
# --- Tests (données statiques) : fit_batch (fonction d'apprentissage) ---
def check_fit_batch():
    model = MLP([2, 1], output_activation="linear", seed=0)
    model.Ws = [np.array([[1.0, -2.0]])]
    model.bs = [np.array([0.0])]
    X = np.array([[1.0, 0.0], [0.0, 1.0]])
    t = np.array([[0.0], [0.0]])
    model.fit_batch(X, t, lr=0.5, epochs=1)

    # simulation à la main (sortie linéaire) :
    #   z = X @ W.T + b = [[1], [-2]]  ;  delta = z - t = [[1], [-2]]
    #   dW = delta.T @ X / 2 = [[0.5, -1]]  ;  db = delta.sum(0) / 2 = -0.5
    #   W -= lr*dW -> [[0.75, -1.5]]  ;  b -= lr*db -> 0.25
    assert np.array_equal(model.Ws[0], np.array([[0.75, -1.5]]))
    assert np.array_equal(model.bs[0], np.array([0.25]))
    print("fit_batch : tests OK")

check_fit_batch()
```

## 2. Jeu de données non linéaire

Deux classes concentriques dans le plan :
- **classe 0** (centre) : disque de rayon ≈ 0.5 ;
- **classe 1** (couronne) : anneau de rayon ≈ 1.0.

Aucune droite ne sépare ces deux classes (le perceptron linéaire du TD01 échoue). Un bruit gaussien radial est ajouté, et on sépare **train/test** pour mesurer la généralisation.

```python
def gen_circles(n=400, noise=0.1, seed=0):
    rng = np.random.default_rng(seed)
    X, t = [], []
    for _ in range(n):
        if rng.random() < 0.5:
            r = 0.5 + rng.normal(0.0, noise)
            label = 0
        else:
            r = 1.0 + rng.normal(0.0, noise)
            label = 1
        theta = rng.uniform(0.0, 2.0 * np.pi)
        X.append([r * np.cos(theta), r * np.sin(theta)])
        t.append(label)
    return np.array(X), np.array(t).reshape(-1, 1)


def plot_points(X, t, ax=None, title=""):
    if ax is None:
        fig, ax = plt.subplots()
    ax.scatter(X[t.ravel() == 0, 0], X[t.ravel() == 0, 1], c="tab:blue", s=20,
               label="classe 0 (centre)", edgecolors="k", linewidths=0.5)
    ax.scatter(X[t.ravel() == 1, 0], X[t.ravel() == 1, 1], c="tab:red", s=20,
               label="classe 1 (couronne)", edgecolors="k", linewidths=0.5)
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.legend()
    ax.grid(True)
    return ax


X, t = gen_circles(n=400, noise=0.1, seed=0)

rng = np.random.default_rng(0)
idx = rng.permutation(len(X))
n_train = int(0.7 * len(X))
X_train, t_train = X[idx[:n_train]], t[idx[:n_train]]
X_test, t_test = X[idx[n_train:]], t[idx[n_train:]]

print("X :", X.shape, "| train :", X_train.shape, "| test :", X_test.shape)
plot_points(X, t, title="Cercles concentriques (bruités)")
```

## 3. k plus proches voisins (k-NN)

**Principe** : pour classer une requête $q$, on cherche ses `k` plus proches voisins dans l'ensemble d'entraînement (distance euclidienne), puis on prend l'étiquette **majoritaire**.

Pas d'apprentissage : tout se fait à la prédiction (modèle *paresseux*).

> **Exercice 1** — Complétez `knn_predict` : calcul des distances puis vote majoritaire.

```python
def knn_predict(X: Mat, t: Mat, Xq: Mat, k: int = 1) -> Mat:
    """Étiquette majoritaire des k plus proches voisins de chaque requête.

    X  : (N, d) points d'entraînement
    t  : (N, 1) étiquettes 0/1
    Xq : (M, d) requêtes
    """
    # BEGIN
    # distances[i, j] = || Xq[i] - X[j] ||  (diffusion (M, N, d))
    diff = Xq[:, None, :] - X[None, :, :]
    dists = np.sqrt((diff ** 2).sum(axis=2))
    # indices des k plus proches voisins pour chaque requête
    idx = np.argsort(dists, axis=1)[:, :k]
    votes = t[idx].reshape(idx.shape[0], k)
    # majorité : moyenne des étiquettes 0/1 (>= 0.5 -> classe 1)
    pred = (votes.mean(axis=1) >= 0.5).astype(float)
    # END
    return pred.reshape(-1, 1)
```

```python
# --- Tests (données statiques) : knn_predict ---
X_t = np.array([[0.0, 0.0], [2.0, 0.0], [0.0, 2.0]])
t_ref = np.array([[0.0], [1.0], [1.0]])

# k=1 : chaque point est son propre plus proche voisin
pred = knn_predict(X_t, t_ref, X_t, k=1)
assert pred.shape == (3, 1)
assert pred.ravel().tolist() == [0.0, 1.0, 1.0]

# k=1 : requêtes déplacées vers un voisin connu
Xq = np.array([[0.1, 0.0], [1.9, 0.0]])
assert knn_predict(X_t, t_ref, Xq, k=1).ravel().tolist() == [0.0, 1.0]

# k=3 : vote majoritaire (1 voisin classe 0, 2 voisins classe 1)
assert knn_predict(X_t, t_ref, np.array([[0.0, 0.0]]), k=3).ravel().tolist() == [1.0]

print("knn_predict : tests OK")
```

```python
def accuracy(t_true, y_pred):
    return float(np.mean(t_true.ravel() == y_pred.ravel()))


def plot_decision(predict_fn, X, t, ax=None, title=""):
    if ax is None:
        fig, ax = plt.subplots()
    x1 = np.linspace(X[:, 0].min() - 0.2, X[:, 0].max() + 0.2, 120)
    x2 = np.linspace(X[:, 1].min() - 0.2, X[:, 1].max() + 0.2, 120)
    xx, yy = np.meshgrid(x1, x2)
    grid = np.c_[xx.ravel(), yy.ravel()]
    Z = predict_fn(grid).reshape(xx.shape)
    ax.contourf(xx, yy, Z, levels=[-0.5, 0.5, 1.5], colors=["tab:blue", "tab:red"], alpha=0.3)
    ax.scatter(X[t.ravel() == 0, 0], X[t.ravel() == 0, 1], c="tab:blue", s=15, edgecolors="k")
    ax.scatter(X[t.ravel() == 1, 0], X[t.ravel() == 1, 1], c="tab:red", s=15, edgecolors="k")
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.grid(False)
    return ax
```

Frontière de décision et précision pour quelques valeurs de `k`.

```python
ks = [1, 3, 10]

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
for ax, k in zip(axes, ks):
    plot_decision(lambda g, k=k: knn_predict(X_train, t_train, g, k=k),
                  X_train, t_train, ax=ax, title=f"k-NN (k={k})")
    acc_tr = accuracy(t_train, knn_predict(X_train, t_train, X_train, k=k))
    acc_te = accuracy(t_test, knn_predict(X_train, t_train, X_test, k=k))
    print(f"k={k}: précision train={acc_tr:.3f}  test={acc_te:.3f}")
plt.tight_layout()
```

## 4. Réseau RBF (fonctions à base radiale)

**Principe** : on place un noyau gaussien sur **chaque point d'entraînement** (les centres $c_j$) :

$$φ_j(x) = \exp\!\left(-\frac{\|x - c_j\|^2}{2σ^2}\right), \qquad c_j = x_j^{(\text{train})}$$

La sortie est un neurone linéaire + sigmoïde sur ces activations :

$$s(x) = σ\!\left(w \cdot φ(x) + b\right), \qquad \hat y = [s(x) > 0.5]$$

Contrairement à k-NN qui ne garde que les `k` plus proches (vote dur), le RBF fait voter **tous** les centres, pondérés par $φ_j(x)$ qui décroît avec la distance (pondération douce). La largeur `σ` joue un rôle analogue à `k` : petit `σ` = voisinage étroit (sur-apprend), grand `σ` = voisinage large (lisse).

> **Exercice 2** — Complétez `rbf_kernel` (noyau gaussien) et `Phi` (matrice des activations).
> **Exercice 3** — Assemblez le RBF pour `σ = 0.5` : centres = points d'entraînement, entraînez `MLP([m, 1], sigmoid)` sur `Phi(X_train)`, prédisez au seuil 0.5.

```python
def rbf_kernel(x: Vec, c: Vec, sigma: float) -> float:
    """Noyau gaussien (fonction de base radiale) : phi_j(x) = exp(-||x - c||^2 / (2 sigma^2))."""
    # BEGIN
    return np.exp(-np.sum((x - c) ** 2) / (2.0 * sigma ** 2))
    # END


def Phi(X: Mat, centers: Mat, sigma: float) -> Mat:
    """Matrice des activations : Phi[i, j] = rbf_kernel(X[i], centers[j]).

    X       : (N, d) points à transformer
    centers : (m, d) centres (= points d'entraînement ici)
    """
    # BEGIN
    # diffusion (N, m, d) : toutes les différences X[i] - centers[j]
    diff = X[:, None, :] - centers[None, :, :]
    sq = (diff ** 2).sum(axis=2)          # (N, m) distances au carré
    return np.exp(-sq / (2.0 * sigma ** 2))
    # END
```

```python
# --- Tests (données statiques) : rbf_kernel et Phi ---
# distance nulle -> noyau = 1
assert rbf_kernel(np.array([0.0, 0.0]), np.array([0.0, 0.0]), 1.0) == 1.0

# ||x - c|| = 1, sigma = 1 -> exp(-1/2)
np.testing.assert_allclose(
    rbf_kernel(np.array([0.0, 0.0]), np.array([1.0, 0.0]), 1.0), np.exp(-0.5)
)

# sigma plus grand -> noyau plus large (exp(-1 / (2 * 100)))
np.testing.assert_allclose(
    rbf_kernel(np.array([0.0, 0.0]), np.array([1.0, 0.0]), 10.0), np.exp(-0.005)
)

X_t = np.array([[0.0, 0.0], [1.0, 0.0]])
C_t = np.array([[0.0, 0.0], [1.0, 0.0]])
P = Phi(X_t, C_t, 1.0)
assert P.shape == (2, 2)
# diagonale = 1 (x == centre)
assert P[0, 0] == 1.0 and P[1, 1] == 1.0
# hors diagonale = exp(-1/2), et symétrie
np.testing.assert_allclose(P[0, 1], np.exp(-0.5))
np.testing.assert_allclose(P[0, 1], P[1, 0])

print("rbf_kernel / Phi : tests OK")
```

```python
# Assemblage du RBF pour sigma = 0.5 (Exercice 3)

# BEGIN
sigma = 0.5
centers = X_train
Phi_train = Phi(X_train, centers, sigma)
Phi_test = Phi(X_test, centers, sigma)

model = MLP([centers.shape[0], 1], output_activation="sigmoid", seed=0)
model.fit_batch(Phi_train, t_train, lr=0.5, epochs=2000)

pred_train = (model.predict(Phi_train) > 0.5).astype(float)
pred_test = (model.predict(Phi_test) > 0.5).astype(float)
# END

print("RBF σ=0.5 : précision train =", round(accuracy(t_train, pred_train), 3),
      " test =", round(accuracy(t_test, pred_test), 3))
```

```python
def train_rbf(X_train, t_train, sigma, activation="sigmoid", lr=0.5, epochs=2000):
    """Construit Phi (centres = points d'entraînement) et entraîne la sortie."""
    centers = X_train
    Phi_train = Phi(X_train, centers, sigma)
    model = MLP([centers.shape[0], 1], output_activation=activation, seed=0)
    model.fit_batch(Phi_train, t_train, lr=lr, epochs=epochs)
    return model, centers


def predict_rbf(model, centers, Xq, sigma, threshold=0.5):
    """Prédiction binaire du RBF (sortie > threshold)."""
    Phi_q = Phi(Xq, centers, sigma)
    return (model.predict(Phi_q) > threshold).astype(float)
```

Frontières côte à côte : ligne du haut = k-NN (`k=1,3,10`), ligne du bas = RBF (`σ=0.1,0.5,1.0`).

## 6. Comparaison des fonctions d'activation de sortie

Jusqu'ici, la sortie du RBF est un neurone **sigmoïde** : $s(x) = σ(w·φ(x) + b)$, et on décide au seuil `0.5`. La classe `MLP` sait aussi produire une sortie `tanh` ou `linear`.

Les trois activations sont **monotones** en $z = w·φ(x) + b$ : la frontière de décision (le lieu où $z = 0$) est donc *théoriquement* identique. Ce qui change, c'est la **carte d'activation** — la valeur brute de la sortie sur le plan :

- `sigmoid` : sortie dans $(0, 1)$, sature vers `0` et `1` → seuil `0.5`.
- `tanh` : sortie dans $(-1, 1)$, symétrique autour de `0` → seuil `0`.
- `linear` : sortie **non bornée**, pas de saturation → seuil `0` ; elle diverge facilement (il faut un `lr` plus petit) et sature moins proprement.

> **Exercice 5** — Entraînez un RBF pour chacune des trois activations de sortie, puis tracez côte à côte la **carte d'activation** (valeur brute, ligne du haut) et la **frontière de décision** (seuil adapté, ligne du bas). Utilisez `lr=0.5` pour `sigmoid`, `lr=1.0` pour `tanh` et `lr=0.02` pour `linear`. Que remarquez-vous sur la forme des zones d'activation ?

```python
def plot_activation(predict_raw, X, t, ax=None, title="", threshold=None):
    if ax is None:
        fig, ax = plt.subplots()
    x1 = np.linspace(X[:, 0].min() - 0.2, X[:, 0].max() + 0.2, 120)
    x2 = np.linspace(X[:, 1].min() - 0.2, X[:, 1].max() + 0.2, 120)
    xx, yy = np.meshgrid(x1, x2)
    grid = np.c_[xx.ravel(), yy.ravel()]
    Z = predict_raw(grid).reshape(xx.shape)
    ax.contourf(xx, yy, Z, levels=60, cmap="coolwarm")
    if threshold is not None:
        ax.contour(xx, yy, Z, levels=[threshold], colors="k", linewidths=1.5)
    ax.scatter(X[t.ravel() == 0, 0], X[t.ravel() == 0, 1], c="tab:blue", s=15, edgecolors="k")
    ax.scatter(X[t.ravel() == 1, 0], X[t.ravel() == 1, 1], c="tab:red", s=15, edgecolors="k")
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.grid(False)
    return ax
```

```python
sigma = 0.5
activations = ["sigmoid", "tanh", "linear"]
thresholds = {"sigmoid": 0.5, "tanh": 0.0, "linear": 0.0}
lrs = {"sigmoid": 0.5, "tanh": 1.0, "linear": 0.02}

# BEGIN
models = {}
for act in activations:
    model, centers = train_rbf(X_train, t_train, sigma, activation=act, lr=lrs[act])
    models[act] = (model, centers)

fig, axes = plt.subplots(2, 3, figsize=(15, 9))

for ax, act in zip(axes[0], activations):
    model, centers = models[act]
    plot_activation(lambda g, m=model, c=centers, s=sigma: m.predict(Phi(g, c, s)),
                    X_train, t_train, ax=ax, title=f"carte d'activation ({act})",
                    threshold=thresholds[act])

for ax, act in zip(axes[1], activations):
    model, centers = models[act]
    plot_decision(lambda g, m=model, c=centers, s=sigma, t=thresholds[act]:
                  (m.predict(Phi(g, c, s)) > t).astype(float),
                  X_train, t_train, ax=ax, title=f"frontière ({act})")
    acc_tr = accuracy(t_train, predict_rbf(model, centers, X_train, sigma, threshold=thresholds[act]))
    acc_te = accuracy(t_test, predict_rbf(model, centers, X_test, sigma, threshold=thresholds[act]))
    print(f"{act}: train={acc_tr:.3f}  test={acc_te:.3f}")

plt.tight_layout()
# END
```

## 7. À retenir

- **k-NN** est *paresseux* : tout le coût est à la prédiction (il faut balayer tout le jeu d'entraînement). Le RBF, une fois entraîné, ne stocke que ses centres et ses poids.
- **k-NN** = sélection **dure** des `k` voisins ; **RBF** = pondération **douce** de tous les voisins par un noyau gaussien. On peut voir le RBF comme un k-NN « lissé ».
- `k` et `σ` contrôlent le **lissage** : trop petit → frontière dentelée (sur-apprentissage) ; trop grand → frontière trop lisse (sous-apprentissage).
- La sortie du RBF est un neurone dont l'**activation** peut être `sigmoid` (sortie `(0,1)`, seuil `0.5`), `tanh` (sortie `(-1,1)`, seuil `0`) ou `linear` (non bornée). Le choix de l'activation fixe le **seuil** de décision ; une activation **bornée et saturante** (`sigmoid`/`tanh`) est plus stable pour classer `{0,1}`.
- Le RBF fait le pont entre k-NN et le perceptron : sa sortie est un neurone linéaire entraîné sur les activations radiales.
