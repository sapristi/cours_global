# Géométrie

Ce TD consiste à écrire des fonctions qui construisent des formes géométriques en « dessin ASCII » (avec des `*` et des espaces).

## Convention commune

Chaque fonction **renvoie** le dessin sous forme d'**une seule chaîne** de caractères (elle n'affiche rien) :

- les lignes du dessin sont séparées par `"\n"` (utiliser `"\n".join(lignes)`) ;
- il n'y a pas de `"\n"` après la dernière ligne ;
- aucune ligne ne se termine par un espace : on n'écrit les espaces que s'ils sont suivis d'une `*` plus loin sur la ligne.

La taille de la figure est passée en paramètres de la fonction. Chaque exercice est suivi d'une cellule de vérification, puis d'une cellule qui affiche le dessin obtenu.

## Exemple

La fonction `affiche_ligne` affiche une ligne de `n` étoiles ; la cellule suivante l'appelle deux fois.

````python
def affiche_ligne(n: int):
    print("*" * n)


affiche_ligne(5)
affiche_ligne(10)
````

On peut aussi construire une ligne en combinant des étoiles et des espaces, ici trois étoiles séparées par cinq espaces :

````python
print("*" + " " * 5 + "*" + " " * 5 + "*")
````

### 1.1 Rectangle

<fieldset style="border: 3px solid #1976d2; border-radius: 8px; padding: 0 14px; margin: 12px 0">
<legend style="color: #1976d2; font-weight: bold; font-size: 1.5em; padding: 0 6px">✏️ Exercice 1</legend>

Complétez `rectangle(largeur, hauteur)` : elle renvoie un rectangle plein de `largeur` colonnes et `hauteur` lignes.

Exemple pour `largeur=5` et `hauteur=3` :

```
*****
*****
*****
```

</fieldset>

````python
def rectangle(largeur: int, hauteur: int) -> str:
    # BEGIN
    # raise NotImplementedError
    ## Indice : `"*" * largeur` donne une ligne.
    lignes = ["*" * largeur for _ in range(hauteur)]
    return "\n".join(lignes)
    # END
````

#### Vérification de `rectangle`

````python
assert rectangle(5, 3) == '*****\n*****\n*****', "rectangle incorrect"
print("rectangle — OK")
````

Après la vérification, la cellule suivante affiche le dessin produit par `rectangle`.

````python
print(rectangle(5, 3))
````

### 1.2 Triangle

<fieldset style="border: 3px solid #1976d2; border-radius: 8px; padding: 0 14px; margin: 12px 0">
<legend style="color: #1976d2; font-weight: bold; font-size: 1.5em; padding: 0 6px">✏️ Exercice 2</legend>

Complétez `triangle(hauteur)` : elle renvoie un triangle rectangle de `hauteur` lignes, aligné à gauche, dont la ligne `i` (en partant de 1) contient `i` étoiles.

Exemple pour `hauteur=4` :

```
*
**
***
****
```

</fieldset>

````python
def triangle(hauteur: int) -> str:
    # BEGIN
    # raise NotImplementedError
    ## Indice : boucler sur `i` de 1 à `hauteur` inclus.
    lignes = ["*" * i for i in range(1, hauteur + 1)]
    return "\n".join(lignes)
    # END
````

#### Vérification de `triangle`

````python
assert triangle(4) == '*\n**\n***\n****', "triangle incorrect"
print("triangle — OK")
````

Après la vérification, la cellule suivante affiche le dessin produit par `triangle`.

````python
print(triangle(4))
````

### 1.3 Autre triangle

<fieldset style="border: 3px solid #1976d2; border-radius: 8px; padding: 0 14px; margin: 12px 0">
<legend style="color: #1976d2; font-weight: bold; font-size: 1.5em; padding: 0 6px">✏️ Exercice 3</legend>

Complétez `autre_triangle(hauteur)` : elle renvoie un triangle rectangle de `hauteur` lignes, aligné à droite. La ligne `i` (en partant de 1) contient `hauteur - i` espaces suivis de `i` étoiles.

Exemple pour `hauteur=4` :

```
   *
  **
 ***
****
```

</fieldset>

````python
def autre_triangle(hauteur: int) -> str:
    # BEGIN
    # raise NotImplementedError
    ## Indice : sur la ligne `i`, `" " * (hauteur - i) + "*" * i`.
    lignes = [" " * (hauteur - i) + "*" * i for i in range(1, hauteur + 1)]
    return "\n".join(lignes)
    # END
````

#### Vérification de `autre_triangle`

````python
assert autre_triangle(4) == '   *\n  **\n ***\n****', "autre_triangle incorrect"
print("autre_triangle — OK")
````

Après la vérification, la cellule suivante affiche le dessin produit par `autre_triangle`.

````python
print(autre_triangle(4))
````

### 1.4 Diamant

<fieldset style="border: 3px solid #1976d2; border-radius: 8px; padding: 0 14px; margin: 12px 0">
<legend style="color: #1976d2; font-weight: bold; font-size: 1.5em; padding: 0 6px">✏️ Exercice 4</legend>

Complétez `diamant(n)` : elle renvoie un losange de `2 * n - 1` lignes et de largeur `2 * n - 1`. Les `n` premières lignes sont centrées et contiennent `1`, `3`, `5`… étoiles (la ligne `i`, en partant de 1, en contient `2 * i - 1` et est précédée de `n - i` espaces) ; les `n - 1` lignes suivantes redescendent symétriquement.

Exemple pour `n=4` :

```
   *
  ***
 *****
*******
 *****
  ***
   *
```

</fieldset>

````python
def diamant(n: int) -> str:
    # BEGIN
    # raise NotImplementedError
    ## Indice : construire la moitié haute, puis ajouter la moitié haute inversée sans sa dernière ligne.
    haut = [" " * (n - i) + "*" * (2 * i - 1) for i in range(1, n + 1)]
    lignes = haut + haut[-2::-1]
    return "\n".join(lignes)
    # END
````

#### Vérification de `diamant`

````python
assert diamant(4) == '   *\n  ***\n *****\n*******\n *****\n  ***\n   *', "diamant incorrect"
print("diamant — OK")
````

Après la vérification, la cellule suivante affiche le dessin produit par `diamant`.

````python
print(diamant(4))
````

### 1.5 Rectangle creux

<fieldset style="border: 3px solid #1976d2; border-radius: 8px; padding: 0 14px; margin: 12px 0">
<legend style="color: #1976d2; font-weight: bold; font-size: 1.5em; padding: 0 6px">✏️ Exercice 5</legend>

Complétez `rectangle_creux(largeur, hauteur)` : elle renvoie le contour d'un rectangle de `largeur` colonnes et `hauteur` lignes (au moins 2 chacune). La première et la dernière ligne sont pleines d'étoiles ; les lignes intermédiaires ont une étoile à chaque extrémité et des espaces entre les deux.

Exemple pour `largeur=11` et `hauteur=4` :

```
***********
*         *
*         *
***********
```

</fieldset>

````python
def rectangle_creux(largeur: int, hauteur: int) -> str:
    # BEGIN
    # raise NotImplementedError
    ## Indice : ligne du milieu = `"*" + " " * (largeur - 2) + "*"`.
    pleine = "*" * largeur
    milieu = "*" + " " * (largeur - 2) + "*"
    lignes = [pleine] + [milieu] * (hauteur - 2) + [pleine]
    return "\n".join(lignes)
    # END
````

#### Vérification de `rectangle_creux`

````python
assert rectangle_creux(11, 4) == '***********\n*         *\n*         *\n***********', "rectangle_creux incorrect"
print("rectangle_creux — OK")
````

Après la vérification, la cellule suivante affiche le dessin produit par `rectangle_creux`.

````python
print(rectangle_creux(11, 4))
````

### 1.6 Rayures

<fieldset style="border: 3px solid #1976d2; border-radius: 8px; padding: 0 14px; margin: 12px 0">
<legend style="color: #1976d2; font-weight: bold; font-size: 1.5em; padding: 0 6px">✏️ Exercice 6</legend>

Complétez `rayures(largeur, hauteur)` : elle renvoie des rayures diagonales dans un rectangle de `largeur` colonnes et `hauteur` lignes. Les colonnes sont numérotées à partir de 0 et les lignes aussi (`ligne` = numéro de la ligne). Une étoile est placée dans la colonne `c` si `(c + ligne) % 4 == 0`, sinon c'est un espace. Les espaces en fin de ligne sont supprimés.

Exemple pour `largeur=17` et `hauteur=4` :

```
*   *   *   *   *
   *   *   *   *
  *   *   *   *
 *   *   *   *
```

</fieldset>

````python
def rayures(largeur: int, hauteur: int) -> str:
    # BEGIN
    # raise NotImplementedError
    ## Indice : construire chaque ligne caractère par caractère, puis utiliser `.rstrip()`.
    lignes = []
    for ligne in range(hauteur):
        texte = ""
        for c in range(largeur):
            texte += "*" if (c + ligne) % 4 == 0 else " "
        lignes.append(texte.rstrip())
    return "\n".join(lignes)
    # END
````

#### Vérification de `rayures`

````python
assert rayures(17, 4) == '*   *   *   *   *\n   *   *   *   *\n  *   *   *   *\n *   *   *   *', "rayures incorrect"
print("rayures — OK")
````

Après la vérification, la cellule suivante affiche le dessin produit par `rayures`.

````python
print(rayures(17, 4))
````

### 1.7 Croix

<fieldset style="border: 3px solid #1976d2; border-radius: 8px; padding: 0 14px; margin: 12px 0">
<legend style="color: #1976d2; font-weight: bold; font-size: 1.5em; padding: 0 6px">✏️ Exercice 7</legend>

Complétez `croix(taille)` : elle renvoie une croix (les deux diagonales) dans un carré de `taille` lignes et `taille` colonnes (`taille` impair). Sur la ligne `i` (à partir de 0), les étoiles sont dans les colonnes `i` et `taille - 1 - i` ; les espaces en fin de ligne sont supprimés.

Exemple pour `taille=5` :

```
*   *
 * *
  *
 * *
*   *
```

</fieldset>

````python
def croix(taille: int) -> str:
    # BEGIN
    # raise NotImplementedError
    ## Indice : sur la ligne du milieu, les deux colonnes coïncident.
    lignes = []
    for i in range(taille):
        texte = ""
        for c in range(taille):
            texte += "*" if c == i or c == taille - 1 - i else " "
        lignes.append(texte.rstrip())
    return "\n".join(lignes)
    # END
````

#### Vérification de `croix`

````python
assert croix(5) == '*   *\n * *\n  *\n * *\n*   *', "croix incorrect"
print("croix — OK")
````

Après la vérification, la cellule suivante affiche le dessin produit par `croix`.

````python
print(croix(5))
````
